### Langkah 1: Bersihkan 2 file BPS (long)

### Langkah 2: Seragamkan nama wilayah

### Langkah 3: Ubah 5 file wide menjadi long

Fungsi `melt` "melelehkan" kolom 2022, 2023, 2024 menjadi baris. Di Excel, ini sama dengan fitur **Unpivot** di Power Query.

Karena langkahnya sama untuk 5 file, kita bungkus dalam satu fungsi sederhana biar ga bolak balik nulis baris code yang sama.

```python
def ubah_ke_long(tabel, nama_kolom):
    hasil = tabel.melt(id_vars="Kabupaten/Kota", var_name="tahun", value_name=nama_kolom)
    hasil = hasil.rename(columns={"Kabupaten/Kota": "kabupaten_kota"})
    hasil["tahun"] = hasil["tahun"].astype(int)    # "2022" (teks) -> 2022 (angka)
    return hasil

# Contoh hasil untuk satu file
contoh = ubah_ke_long(pengangguran, "pengangguran")
print(pengangguran.shape, "->", contoh.shape)
contoh[contoh["kabupaten_kota"] == "Aceh Barat"]
```

```python
ipm_long = ubah_ke_long(ipm, "ipm")
pengangguran_long = ubah_ke_long(pengangguran, "pengangguran")
putus_long = ubah_ke_long(putus_sekolah, "putus_sekolah")
kepadatan_long = ubah_ke_long(kepadatan, "kepadatan")
umr_long = ubah_ke_long(umr, "umr")

print(ipm_long.shape, pengangguran_long.shape, putus_long.shape, kepadatan_long.shape, umr_long.shape)
```

File pendapatan tidak diubah karena tidak dipakai (keputusan A4).

### Langkah 4: Buat tabel referensi provinsi
Di file kemiskinan, setiap baris provinsi (HURUF BESAR) selalu diikuti kabupaten/kota di dalamnya.
Triknya: salin nama provinsi ke kolom baru, lalu **isi ke bawah** (`ffill`, sama seperti *Fill Down* di Excel)
sampai bertemu provinsi berikutnya. Kita pakai data **2024** supaya kabupaten di Papua masuk ke provinsi barunya.

```python
ref = miskin[miskin["tahun"] == 2024][["kabupaten_kota"]].copy()

adalah_provinsi = ref["kabupaten_kota"].str.isupper()
ref["provinsi"] = ref["kabupaten_kota"].where(adalah_provinsi)   # isi nama provinsi hanya di baris provinsi
ref["provinsi"] = ref["provinsi"].ffill()                        # Fill Down
ref = ref[~adalah_provinsi]                                      # buang baris provinsinya

n_kab = len(ref)
print("Jumlah kab/kota:", n_kab, "| jumlah provinsi:", ref["provinsi"].nunique())
ref[ref["kabupaten_kota"].isin(["Asmat", "Merauke", "Kota Sorong", "Kota Jayapura"])]
```

# A6. Mengintegrasikan Data

Tujuan: menggabungkan semua tabel menjadi **satu tabel** dengan kunci `kabupaten_kota` + `tahun`.

`merge` di pandas sama dengan **VLOOKUP / XLOOKUP** di Excel: mencari baris yang kuncinya sama, lalu menempelkan kolomnya.
Tabel dasar adalah `ref` (daftar kab/kota resmi), digandakan untuk 3 tahun.

```python
# Tabel dasar: setiap kab/kota x 3 tahun
dasar = ref.merge(pd.DataFrame({"tahun": [2022, 2023, 2024]}), how="cross")
print("Tabel dasar:", dasar.shape)

# Tempelkan indikator satu per satu (how="left": semua baris tabel dasar dipertahankan)
data = dasar.merge(miskin, on=["kabupaten_kota", "tahun"], how="left")
data = data.merge(pengeluaran, on=["kabupaten_kota", "tahun"], how="left")
data = data.merge(ipm_long, on=["kabupaten_kota", "tahun"], how="left")
data = data.merge(pengangguran_long, on=["kabupaten_kota", "tahun"], how="left")
data = data.merge(putus_long, on=["kabupaten_kota", "tahun"], how="left")
data = data.merge(kepadatan_long, on=["kabupaten_kota", "tahun"], how="left")
data = data.merge(umr_long, on=["kabupaten_kota", "tahun"], how="left")

n_data = len(data)
print("Hasil gabungan:", data.shape)
```

```python
# Cek integritas hasil gabungan
print("Jumlah baris = kab/kota x 3 tahun?", n_data == n_kab * 3)
print("Baris dobel                       :", data.duplicated(subset=["kabupaten_kota", "tahun"]).sum())
print("Nilai kosong per kolom:")
print(data.isna().sum())
```

**Tidak ada nilai kosong dan tidak ada baris dobel.** Ini bukti pembersihan nama dan baris dobel di A5 berhasil.

### Rekayasa fitur (buat kolom baru) --- OPSIONAL alias BOLEH DI-SKIP
- `jenis_wilayah` - "Kota" jika nama diawali `Kota `, selain itu "Kabupaten" - Untuk membandingkan kota vs kabupaten
- `kategori_kemiskinan` - Menggunakan `pd.qcut` membagi P0 menjadi 3 kelompok sama besar: Rendah/Sedang/Tinggi - Berguna untuk pewarnaan peta dan filter dashboard
-  `kategori_ipm` - Menggunakan `pd.cut` dengan batas kategori BPS (60, 70, 80) - Berguna untuk filter dashboard

```python
data["jenis_wilayah"] = "Kabupaten"
data.loc[data["kabupaten_kota"].str.startswith("Kota "), "jenis_wilayah"] = "Kota"

data["kategori_kemiskinan"] = pd.qcut(data["penduduk_miskin"], q=3, labels=["Rendah", "Sedang", "Tinggi"])

data["kategori_ipm"] = pd.cut(data["ipm"], bins=[0, 60, 70, 80, 101], right=False,
                              labels=["Rendah", "Sedang", "Tinggi", "Sangat Tinggi"])

# Batas kategori kemiskinan (untuk dicatat di presentasi)
batas = data["penduduk_miskin"].quantile([1/3, 2/3]).round(2)
batas_rendah = batas.iloc[0]
batas_tinggi = batas.iloc[1]
print("Rendah: P0 <", batas_rendah, "| Sedang:", batas_rendah, "-", batas_tinggi, "| Tinggi: P0 >", batas_tinggi)
print(data[data["tahun"] == 2024]["jenis_wilayah"].value_counts())
```
